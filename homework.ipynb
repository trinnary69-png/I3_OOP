```python
#4.5 practice 5: Temperature Converter
class Temperature:
    @staticmethod
    def celsius_to_fahrenheit(celsius):
        return (9 / 5) * celsius + 32

    @staticmethod
    def fahrenheit_to_celsius(fahrenheit):
        return (5 / 9) * (fahrenheit - 32)

    @staticmethod
    def celsius_to_kelvin(celsius):
        return celsius + 273.15


celsius = 25
print(f"Celsius: {celsius}")
print(f"Fahrenheit: {Temperature.celsius_to_fahrenheit(celsius)}")
print(f"Kelvin: {Temperature.celsius_to_kelvin(celsius)}")
```

```python
#4.6 practice 6: Product Management System 
class Product:
    store_name = "ITC Online Store"
    product_count = 0

    def __init__(self, name, price, quantity):
        self.name = name
        self.price = price
        self.quantity = quantity
        Product.product_count += 1

    def display_product(self):
        print(f"Name: {self.name}")
        print(f"Price: ${self.price}")
        print(f"Quantity: {self.quantity}")
        print(f"Total Value: ${self.total_value()}")

    def total_value(self):
        return self.price * self.quantity

    @classmethod
    def get_product_count(cls):
        return cls.product_count

    @staticmethod
    def is_valid_price(price):
        return price > 0


products = [
    Product("Laptop", 800, 5),
    Product("Mouse", 15, 20),
    Product("Keyboard", 30, 10),
]

print(f"---------- {Product.store_name.upper()} ----------")
print()

inventory_total = 0
for i, product in enumerate(products, start=1):
    print(f"Product {i}")
    product.display_product()
    print()
    inventory_total += product.total_value()

print(f"Total Products: {Product.get_product_count()}")
print(f"Total Inventory Value: ${inventory_total}")
```
