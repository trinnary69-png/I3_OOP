```python

class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.__price = price

    def get_price(self):
        return self.__price

    def set_price(self, price):
        if price > 0:
            self.__price = price
        else:
            print("Invalid price!")

class EBook(Book):
    def __init__(self, title, author, price, file_size):
        super().__init__(title, author, price)
        self.file_size = file_size

book1 = Book("Python Programming", "John", 15)

print("Book Information")
print("Title:", book1.title)
print("Author:", book1.author)
print("Price:", book1.get_price())

ebook1 = EBook("Python Basics", "David", 20, "5 MB")

print("\nEBook Information")
print("Title:", ebook1.title)
print("Author:", ebook1.author)
print("Price:", ebook1.get_price())
print("File Size:", ebook1.file_size)
```
